W5A23 Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.

W5A24 Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.


```python
import math
x1, y1=map(float,input().split())
x2, y2=map(float,input().split())
x3, y3=map(float,input().split())
def are_collinear(x1,y1,x2,y2,x3,y3):
    AB=math.sqrt((x2-x1)**2+(y2-y1)**2)
    AC=math.sqrt((x3-x1)**2+(y3-y1)**2)
    BC=math.sqrt((x3-x2)**2+(y3-y2)**2)
    p=(AB+AC+BC)/2
    s=math.sqrt(p*(p-AB)*(p-AC)*(p-BC))
    if s>0:
        return "không thẳng hàng"
    elif s==0:
        return "thẳng hàng"
print(are_collinear(x1,y1,x2,y2,x3,y3))

```

Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3),
(0,3)], p = (2,1)


```python
def point_on_segment(a, b, p):
    x1, y1 = a
    x2, y2 = b
    x, y = p

    cross = (x - x1) * (y2 - y1) - (y - y1) * (x2 - x1)
    if abs(cross) > 1e-9:
        return False

    return (min(x1, x2) <= x <= max(x1, x2) and
            min(y1, y2) <= y <= max(y1, y2))
def point_in_quadrilateral(quad, p):
    n = len(quad)

    for i in range(n):
        if point_on_segment(quad[i], quad[(i + 1) % n], p):
            return True

    x, y = p
    inside = False

    for i in range(n):
        x1, y1 = quad[i]
        x2, y2 = quad[(i + 1) % n]

        if ((y1 > y) != (y2 > y)):
            x_intersect = x1 + (y - y1) * (x2 - x1) / (y2 - y1)
            if x < x_intersect:
                inside = not inside

    return inside
quad = eval(input("Nhập tứ giác: "))
p = eval(input("Nhập điểm p: "))
print(point_in_quadrilateral(quad, p))
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3)

```python
import math
x1, y1=map(int,input().split())
x2, y2=map(int,input().split())
x3, y3=map(int,input().split())
def tinhgocABC(x1,y1,x2,y2,x3,y3):
    AB=math.sqrt((x2-x1)**2+(y2-y1)**2)
    BC=math.sqrt((x3-x2)**2+(y3-y2)**2)
    AC=math.sqrt((x3-x1)**2+(y3-y1)**2)
    cos_ABC=(AB**2+BC**2-AC**2)/(2*AB*BC)
    goctheoradian=math.acos(cos_ABC)
    goctheodo=math.degrees(goctheoradian)
    return goctheodo
print(tinhgocABC(x1,y1,x2,y2,x3,y3))
    
```

Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa,
giá km, phí chờ, phụ phí ban đêm và giảm giá

```python
import math 
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
  if car_type==4:
    s=11000
    rate1, rate2=13000, 10500
  elif car_type==7:
    s=12000
    rate1, rate2=14000, 12000
  if 0.8<km<=20:
    s+=(km-0.8)*rate1
  elif km>20:
    s+=19.2*rate1+(km-20)*rate2
  if wait_minutes<=5:
    s+=0
  elif wait_minutes>5:
    tgchotinhphi=math.ceil(((wait_minutes-5)/5))
    s+=tgchotinhphi*2000
  if start_hour in [22,23,0,1,2,3,4,5]:
    s=1.2*s
  if km>30 and start_hour not in [22,23,0,1,2,3,4,5]:
    s=0.9*s
  if s<50000 and km>5:
    s=50000
  return s
car_type, start_hour, km, wait_minutes=map(float,input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau
Nếu n ≤ 0 trả về 0

```python
def sum_1_to_n(n):
    s=0
    for i in range(1,n+1):
        for j in range(len(str(i))):
            if str(i)[j-1]!=str(i)[j]:
                s+=i
    return s
n=int(input())
print(sum_1_to_n(n))
```
